```python
import matplotlib.pyplot as plt
import pandas as pd
from pathlib import Path
import numpy as np

file_path = Path.cwd() / "dataset" / "Amazon_Reviews.csv"


df = pd.read_csv(file_path, engine = 'python')

```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
print(df.isnull().sum() / len(df) * 100)
```

```python
print(df.shape)
```

```python
df = df.dropna().reset_index(drop = True)
print(df.shape)
```

```python
df["raw_date"] = pd.to_datetime(df["Review Date"])
df["year"] = df["raw_date"].dt.year
df["month"] = df["raw_date"].dt.month
df["day"] = df["raw_date"].dt.day

df["rating"] = df["Rating"].str.split().str[1].astype(np.int8)

df["review count"] = df["Review Count"].str.split().str[0].astype(np.int32)

df_clean = df.drop(columns = ["Review Count", "Review Date", "Rating", "Date of Experience", "Profile Link"])
df_clean.to_csv("dataset/dataset_clean.csv", index = False)

```
